# Baseline Operating Patterns

Use this notebook to inspect normal baseline behavior, missingness, and feature groups. Keep transformations and model training in the package.

```python
from pathlib import Path

from sfsa.data.rbc_ashrae1312 import load_scenario
from sfsa.features import classify_columns, missingness_report

BASELINE = Path("../data/raw/01_RBC-ASHRAE1312/Real/Summer/20070820/baseline.csv")
baseline, metadata = load_scenario(BASELINE)
groups = classify_columns(baseline.columns)

print(metadata)
print(groups.as_dict())
missingness_report(baseline).head(20)
```
