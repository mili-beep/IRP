# RBC-ASHRAE1312 Real Data Audit

This notebook inspects the real-only dataset and reports file coverage, scenario metadata, timestamp quality, and baseline/fault pairing. It does not train models.

```python
from pathlib import Path

from sfsa.data.rbc_ashrae1312 import discover_real_files, validate_pairing

DATA_ROOT = Path("../data/raw/01_RBC-ASHRAE1312/Real")
records = discover_real_files(DATA_ROOT)
pairings = validate_pairing(DATA_ROOT)

print(f"Scenario/baseline files: {len(records)}")
print(f"Fault files: {sum(record.is_fault for record in records)}")
print(f"Seasons: {sorted({record.season for record in records})}")
print(f"Missing baselines: {sum(not item['baseline_present'] for item in pairings)}")
print(f"Unaligned pairs: {sum(not item['timestamps_aligned'] for item in pairings)}")
```

```python
from sfsa.data.manifest import build_real_manifest

manifest = build_real_manifest(DATA_ROOT)
print(manifest['validation'])
display(__import__('pandas').DataFrame(manifest['files']).groupby(['season', 'is_fault']).size())
```
